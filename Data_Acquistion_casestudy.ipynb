```python
# Parse XML file
tree = ET.parse("literacy_rate.xml")
root = tree.getroot()

# Extract records
data = []

for record in root:
    row = {}
    for field in record:
        row[field.tag] = field.text
    data.append(row)

# Convert into DataFrame
literacy_rate_df = pd.DataFrame(data)
literacy_rate_df
```

Task 8

Merge all four datasets into one DataFrame.

Hint:

● Use the common column (country)

● Choose an appropriate join method.

```python
print("Population:", population_df.columns.tolist())
print("Internet Users:", internet_users_df.columns.tolist())
print("GDP:", gdp_df.columns.tolist())
print("Literacy Rate:", literacy_rate_df.columns.tolist())
```

```python
display(literacy_rate_df.head(10))
```

```python
print(literacy_rate_df.info())
```

```python
population_df.columns = population_df.columns.str.strip().str.lower()
internet_users_df.columns = internet_users_df.columns.str.strip().str.lower()
gdp_df.columns = gdp_df.columns.str.strip().str.lower()
literacy_rate_df.columns = literacy_rate_df.columns.str.strip().str.lower()
```

```python
population_literacy_df = population_df.merge(
    literacy_rate_df,
    on='population',
    how='inner'
)

display(population_literacy_df.head())
```

```python
final_df = (
    population_literacy_df
    .merge(gdp_df, on='country', how='inner')
    .merge(internet_users_df, on='country', how='inner')
)

display(final_df.head())
```

```python
print("Final Shape:", final_df.shape)
print("Columns:", final_df.columns.tolist())

display(final_df.head())
```

Task 9

Create the following calculated
columns:

1. Internet Penetration (%)
2. GDP Per Capita

Hint:
Round the calculated values to two decimal places.

```python
final_df['Internet Penetration (%)'] = (
    final_df['internet_users'] / final_df['population']
) * 100

final_df['Internet Penetration (%)'] = final_df['Internet Penetration (%)'].round(2)

final_df['gdp Per Capita'] = (
    final_df['gdp'] / final_df['population']
).round(2)
```

```python
#List all countries having a literacy rate greater than 90%.
conn = sqlite3.connect('country_statistics.db')
query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90;
"""
result = pd.read_sql_query(query, conn)
display(result)

conn.close()
```

```python
# List all countries having an internet penetration greater than 70%.
conn = sqlite3.connect('country_statistics.db')

query = """
SELECT country, "Internet Penetration (%)"
FROM country_statistics
WHERE "Internet Penetration (%)" > 70;
"""
result = pd.read_sql_query(query, conn)
display(result)

conn.close()
```

```python
#List the Top 5 countries by GDP per capita.
conn = sqlite3.connect('country_statistics.db')
query = """
SELECT country, "GDP Per Capita"
FROM country_statistics
ORDER BY "GDP Per Capita" DESC
LIMIT 5;
"""
result = pd.read_sql_query(query, conn)
display(result)

conn.close()
```

```python
#Calculate the total population represented in the dataset.
conn = sqlite3.connect('country_statistics.db')

query = """
SELECT SUM(population) AS total_population
FROM country_statistics;
"""
result = pd.read_sql_query(query, conn)
display(result)

conn.close()
```

```python
#Calculate the average internet penetration across all countries.
conn = sqlite3.connect('country_statistics.db')

query = """
SELECT AVG("Internet Penetration (%)") AS average_internet_penetration
FROM country_statistics;
"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
#Identify the country having the highest number of internet users.
conn = sqlite3.connect('country_statistics.db')

query = """
SELECT country, MAX(internet_users) AS highest_internet_users
FROM country_statistics;
"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
# Find the correlation between Literacy Rate and Internet Penetration.
correlation = final_df['literacy_rate'].corr(
    final_df['Internet Penetration (%)']
)

print("Correlation:", round(correlation, 2))
```
